Step 3: Train and Save the Machine Learning Classifier
Here, we use a Random Forest model. It learns from features like genre, language, and energy to classify and recommend songs for different moods.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import joblib
import numpy as np

FEATURES = ['genre', 'language', 'energy']
TARGET = 'mood'

# Optimized Random Forest Classifier with balanced class weights and tuned parameters
model = Pipeline([
    ('encode', ColumnTransformer([
        ('categorical', OneHotEncoder(handle_unknown='ignore', sparse_output=False), FEATURES)
    ])),
    ('classifier', RandomForestClassifier(
        n_estimators=500,
        max_depth=15,
        min_samples_split=2,
        min_samples_leaf=1,
        random_state=42,
        class_weight='balanced'
    ))
])

# Splitting and testing the model on the enlarged, rich dataset
class_counts = df[TARGET].value_counts()
can_split = len(df) >= 50 and class_counts.min() >= 5

if can_split:
    X_train, X_test, y_train, y_test = train_test_split(
        df[FEATURES], df[TARGET], test_size=0.20,
        random_state=42, stratify=df[TARGET]
    )
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print('Hold-out validation accuracy:', round(accuracy_score(y_test, pred), 3))
    print('\nModel Classification Report:')
    print(classification_report(y_test, pred, zero_division=0))
else:
    print('Dataset was insufficient to perform a stratified evaluation.')

# Re-train on full dataset for live app use
model.fit(df[FEATURES], df[TARGET])
MODEL_PATH = DATA_DIR / 'mood_model.joblib'
joblib.dump(model, MODEL_PATH)
print('\nSuccessfully trained and saved the optimized model at:', MODEL_PATH)
```

```python
from flask import Flask, request, render_template_string, jsonify
import threading
import time
import pandas as pd
import joblib
from pathlib import Path

app = Flask('moodtune')

DATA_DIR = Path('/content/moodtune_data')
CSV_PATH = DATA_DIR / 'songs.csv'
df = pd.read_csv(CSV_PATH).fillna('')

FEATURES = ['genre', 'language', 'energy']
MODEL_PATH = DATA_DIR / 'mood_model.joblib'

try:
    model = joblib.load(MODEL_PATH)
    print("ML Model successfully loaded for live suggestions.")
except Exception as e:
    model = None
    print("Fallback to direct rules. Model load issue:", e)

PAGE = r'''
<!doctype html>
<html lang="en">
<head>
    <meta charset="utf-8">
    <meta name="viewport" content="width=device-width,initial-scale=1">
    <title>MoodTune — Smart Music Recommendation System</title>
    <style>
        * { box-sizing: border-box; }
        body {
            margin: 0;
            background: #0c0b0b;
            color: #ffffff;
            font-family: 'Segoe UI', -apple-system, BlinkMacSystemFont, Roboto, sans-serif;
        }
        main {
            max-width: 900px;
            margin: auto;
            padding: 40px 20px;
        }
        .brand {
            font-size: 28px;
            font-weight: 800;
            color: #1DB954;
            display: flex;
            align-items: center;
            gap: 10px;
            margin-bottom: 20px;
        }
        .brand span {
            font-size: 32px;
        }
        .hero {
            text-align: center;
            padding: 20px 0 40px;
        }
        .hero h1 {
            font-size: clamp(32px, 5vw, 48px);
            margin: 10px 0;
            font-weight: 800;
            letter-spacing: -1px;
        }
        .hero p, .muted {
            color: #a7a7a7;
            font-size: 16px;
        }
        .panel, .results {
            background: #121212;
            border: 1px solid #282828;
            border-radius: 12px;
            padding: 24px;
            margin-bottom: 24px;
        }
        .grid {
            display: grid;
            grid-template-columns: repeat(4, 1fr);
            gap: 16px;
        }
        label {
            display: flex;
            flex-direction: column;
            gap: 8px;
            color: #b3b3b3;
            font-size: 13px;
            font-weight: 600;
            text-transform: uppercase;
            letter-spacing: 0.5px;
        }
        select {
            padding: 12px;
            border-radius: 6px;
            background: #181818;
            color: #ffffff;
            border: 1px solid #3e3e3e;
            outline: none;
            font-size: 14px;
            cursor: pointer;
            transition: border-color 0.2s;
        }
        select:focus {
            border-color: #1DB954;
        }
        button {
            width: 100%;
            padding: 14px;
            margin-top: 24px;
            border: 0;
            border-radius: 50px;
            background: #1DB954;
            color: #ffffff;
            font-weight: bold;
            font-size: 16px;
            cursor: pointer;
            transition: background 0.2s, transform 0.1s;
        }
        button:hover {
            background: #1ed760;
            transform: scale(1.01);
        }
        .song {
            display: flex;
            gap: 16px;
            align-items: center;
            padding: 16px 0;
            border-bottom: 1px solid #282828;
        }
        .song:last-child {
            border-bottom: none;
        }
        .cover {
            background: linear-gradient(135deg, #1DB954, #191414);
            width: 50px;
            height: 50px;
            border-radius: 6px;
            display: grid;
            place-items: center;
            font-size: 22px;
            font-weight: bold;
        }
        .info {
            flex: 1;
        }
        .info b {
            display: block;
            font-size: 16px;
            color: #ffffff;
            margin-bottom: 4px;
        }
        .info span {
            display: block;
            color: #b3b3b3;
            font-size: 13px;
        }
        .info small {
            display: block;
            color: #a7a7a7;
            margin-top: 6px;
            font-size: 11px;
            text-transform: uppercase;
            letter-spacing: 0.5px;
        }
        .actions {
            display: flex;
            align-items: center;
            gap: 12px;
        }
        .listen-btn {
            background: #242424;
            color: #ffffff;
            text-decoration: none;
            padding: 8px 16px;
            border-radius: 20px;
            font-size: 12px;
            font-weight: bold;
            border: 1px solid #3e3e3e;
            transition: background 0.2s, border-color 0.2s;
        }
        .listen-btn:hover {
            background: #1DB954;
            border-color: #1DB954;
        }
        footer {
            text-align: center;
            color: #7a7a7a;
            font-size: 13px;
            margin: 40px 0 20px;
        }
        @media (max-width: 768px) {
            .grid { grid-template-columns: 1fr; }
            .song { flex-direction: column; align-items: flex-start; }
            .actions { width: 100%; justify-content: flex-end; }
        }
    </style>
</head>
<body>
<main>
    <div class="brand"><span>♫</span> MoodTune</div>

    <section class="hero">
        <p>SMART MUSIC DISCOVERY PORTAL</p>
        <h1>Ready for your music?</h1>
        <p class="muted">Select your preferred choices simultaneously. All filters are simple, independent, and fast.</p>
    </section>

    <section class="panel">
        <form id="filterForm" method="POST" action="/">
            <div class="grid">
                <label>1. Mood
                    <select name="mood" id="moodSelect" required>
                        <option value="">Select Mood</option>
                        {% for m in moods %}
                        <option value="{{ m }}" {% if selected_mood == m %}selected{% endif %}>{{ m }}</option>
                        {% endfor %}
                    </select>
                </label>

                <label>2. Genre
                    <select name="genre" id="genreSelect">
                        <option value="Any">Any</option>
                        {% for g in genres %}
                        <option value="{{ g }}" {% if selected_genre == g %}selected{% endif %}>{{ g }}</option>
                        {% endfor %}
                    </select>
                </label>

                <label>3. Language
                    <select name="language" id="languageSelect">
                        <option value="Any">Any</option>
                        {% for l in languages %}
                        <option value="{{ l }}" {% if selected_language == l %}selected{% endif %}>{{ l }}</option>
                        {% endfor %}
                    </select>
                </label>

                <label>4. Energy
                    <select name="energy" id="energySelect">
                        <option value="Any">Any</option>
                        {% for e in energies %}
                        <option value="{{ e }}" {% if selected_energy == e %}selected{% endif %}>{{ e }}</option>
                        {% endfor %}
                    </select>
                </label>
            </div>
            <button type="submit">Find my songs →</button>
        </form>
    </section>

    {% if submitted %}
    <section class="results">
        <h2 style="margin-top:0; margin-bottom: 20px; font-size: 20px; border-bottom: 1px solid #282828; padding-bottom: 12px;">
            Recommendations ({{ selected_mood }} Mood · {{ results|length }} songs matched)
        </h2>
        {% if results %}
            {% for s in results %}
            <article class="song">
                <div class="cover">♫</div>
                <div class="info">
                    <b>{{ s.song_name }}</b>
                    <span>{{ s.artist }}</span>
                    <small>{{ s.language }} · {{ s.genre }} · {{ s.energy }} Energy</small>
                </div>
                <div class="actions">
                    <a class="listen-btn" target="_blank" rel="noopener" href="{{ s.full_song_url }}">Play on YouTube ↗</a>
                </div>
            </article>
            {% endfor %}
        {% else %}
            <p class="muted">No exact matches found for this specific combination. Try relaxing your filters.</p>
        {% endif %}
    </section>
    {% endif %}

    <footer>
        MoodTune Music Recommender Project &copy; 2026. All rights reserved.
    </footer>
</main>
</body>
</html>
'''

@app.route('/', methods=['GET', 'POST'])
def home():
    moods = sorted(df['mood'].astype(str).unique().tolist())
    genres = sorted(df['genre'].astype(str).unique().tolist())
    languages = sorted(df['language'].astype(str).unique().tolist())
    energies = sorted(df['energy'].astype(str).unique().tolist())

    results, submitted, selected_mood = [], False, ''
    selected_genre = 'Any'
    selected_language = 'Any'
    selected_energy = 'Any'

    if request.method == 'POST':
        submitted = True
        selected_mood = request.form.get('mood', '')
        selected_genre = request.form.get('genre', 'Any')
        selected_language = request.form.get('language', 'Any')
        selected_energy = request.form.get('energy', 'Any')

        candidates = df.copy()
        if model is not None:
            candidates['predicted_mood'] = model.predict(candidates[FEATURES])
            candidates = candidates[candidates['predicted_mood'] == selected_mood]
        else:
            candidates = candidates[candidates['mood'] == selected_mood]

        if selected_genre != 'Any':
            candidates = candidates[candidates.genre == selected_genre]
        if selected_language != 'Any':
            candidates = candidates[candidates.language == selected_language]
        if selected_energy != 'Any':
            candidates = candidates[candidates.energy == selected_energy]

        results = candidates.head(15).to_dict('records')

    return render_template_string(PAGE, moods=moods, genres=genres, languages=languages, energies=energies,
                                  results=results, submitted=submitted, selected_mood=selected_mood,
                                  selected_genre=selected_genre, selected_language=selected_language,
                                  selected_energy=selected_energy)

def run_flask():
    app.run(host='0.0.0.0', port=5000, debug=False, use_reloader=False)

# Restart thread safely
if not any(t.name == 'moodtune-flask' and t.is_alive() for t in threading.enumerate()):
    server_thread = threading.Thread(target=run_flask, daemon=True, name='moodtune-flask')
    server_thread.start()
    time.sleep(2)
print('Flask server successfully updated and started on port 5000 with simple parallel dropdown options.')
```
